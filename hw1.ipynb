给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9



idea:
实际利润就是 这两个日期之间的所有 profits 的总和。所以就是找一个连续子数组，数组和最大即可。
sum存储累积收益，累积收益最小，就是买入的时候。如果累积收益大于最大利润，则更新最大利润与卖出时间，直至循环结束。
时间复杂度为O(n) 。

````python
# add your code
profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
n = len(profits)

sum = []
max_profit = profits[0]
current_sum = 0
buy = 1
sell = 1
temp_buy = 1

for i in range(n):
    current_sum += profits[i]
    sum.append(current_sum)
    
    if current_sum > max_profit:
        max_profit = current_sum
        buy = temp_buy
        sell = i + 1  # 转为从第1天开始计数

    if current_sum < 0:
        current_sum = 0
        temp_buy = i + 2  # 下一个区间起点（i+1转为从1开始）

print("sum=" , sum)
print("买入日期为",buy,"卖出日期为",sell,"最大利润为",max_profit)

"T(n)=O(n)"
````

1.idea：这里使用快排，是在第5题中使用的快排函数。

````python
# add your code here
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]

def quick_sort(arr, start, end):
    if start >= end:  
        return
    mid = arr[start]  
    left = start  
    right = end  
    while left < right:
        while left <  right and arr[right] >= mid:
            right -= 1
        arr[left] = arr[right]  
        while left < right and arr[left] < mid:
            left += 1
        arr[right] = arr[left]  
    arr[left] = mid  
    quick_sort(arr, start, left - 1) 
    quick_sort(arr, left + 1, end) 
    return arr
# 执行
result = quick_sort(arr, 0, len(arr)-1)
print("前三个最小数是：",result[:3])

# your algorithm time complexity is :
"T(n)=O(nlogn)"
````

2.idea：输出最小值，自然是使用最小堆。
从空堆逐个插入建立最小堆，并调用 k 次 Extract-Min 输出最小的 k 个数。

````python
# add your code here
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
import heapq

def build_min_heap_insert(arr):
    heap = []
    for n in arr:
        # 插入一个元素
        heapq.heappush(heap, n)  
    return heap
 # 提取最小元素
def Extract_min(heap, k):
    result = []  # 用于保存结果
    for _ in range(k):
        if len(heap) > 0:  # 判断堆不为空
            min = heapq.heappop(heap)  # 弹出最小元素
            result.append(min)
    return result

k = 3
# 构建最小堆（插入建堆）
min_heap = build_min_heap_insert(arr)

# 提取前 k 个最小值
min = Extract_min(min_heap, k)

print("前", k, "个最小元素为：", min)

# your algorithm time complexity is :
print("建堆时间复杂度：","T(n)=O(nlogn)")
print("提取时间复杂度：","T(n)=O(klogn)")
print("总体时间复杂度：","T(n)=O(nlogn+klogn)")
````

给定一个包含m个**字符串**的数组A, 其中不同的字符串可能有不同的字符数, 但数组中所有字符串的字符总数为n。设计一个算法在 $O(n)$ 时间内对字符串进行排序, 分析算法设计方案, 计算其时间复杂度, 并基于python编程实现该算法。请注意, 假设字符串只包含"a","b",...,"z", 



举例1：数组A=["a", "da", "bde", "ab", "bc", "abdc", "cdba"], 排序后的数组应该为：['a', 'ab', 'abdc', 'bc', 'bde', 'cdba', 'da']



举例2：数组A=['ab', 'a', 'b', 'abc', 'ba', 'c'], 排序后的数组应该为：['a', 'ab', 'abc', 'b', 'ba', 'c']



举例3：数组A=['aef', 'yzr', 'wr', 'ab', 'bhjc', 'lkabdc', 'pwcdba'],  排序后的数组应该为：['ab', 'aef', 'bhjc', 'lkabdc', 'pwcdba', 'wr', 'yzr']



note：

-  两个字符之间的比较可以考虑比较他们对应的ASCII码值；
- python中可以使用`ord("a")`返回字符 “a”对应的ASCII值

idea:数组中所有字符串的字符总数为n。那就将a到z看成1到26的自然数。
使用ord获取字符的ASCII值，然后减去a的ASCII值，就可映射到1到26的自然数。桶0是''空字符。
从“右往左”处理每一位字符，把字符放进对应的桶中。
字符串中字母多的放在与他头部字母相同的后面，如abc放在ab后面，ab放在a后面。找最大长度，以及从右向左，就是为了先把abc放在后面。

桶方法参考了CSDN上的相关文章，在其基础上进行了学习与修改。

````python
# add your code here
def stable_sort(strings, pos):
    # 27 个桶：0表示缺位，1~26表示 a~z
    buckets = []
    for i in range(27):
        buckets.append([])
    for s in strings:
        # 如果当前字符串太短，没有pos位置的字符，就把他放入桶0
        if pos >= len(s):
            buckets[0].append(s)
        else:
            ch = s[pos]
            bucket_index = ord(ch) - ord('a') + 1
            buckets[bucket_index].append(s)
    result = []
    for i in buckets:
        result.extend(i)
    return result

def sort_strings(A):
    if not A:
        return []
    max_len = max(len(s) for s in A)  # 找出最长字符串的长度
     # 从字符串的最后一位往前排
    for i in reversed(range(max_len)): # i 从 max_len - 1 到 0
        A = stable_sort(A, i)
    return A

A1 = ["a", "da", "bde", "ab", "bc", "abdc", "cdba"]
A2 = ['ab', 'a', 'b', 'abc', 'ba', 'c']
A3 = ['aef', 'yzr', 'wr', 'ab', 'bhjc', 'lkabdc', 'pwcdba']

print("示例1排序结果：", sort_strings(A1))
print("示例2排序结果：", sort_strings(A2))
print("示例3排序结果：", sort_strings(A3))

# your algorithm time complexity is :
"T(n)=O(n)"
````
