## 问题 1  

> 给定一个已排序的链表的头 `head` ， *删除所有重复的元素，使每个元素只出现一次* 。返回 *已排序的链表* 。链表的类如下所示：

```python
class NodeList:
    def __init__(self, val=None, right=None):
        self.val   = val
        self.right = right
```

输入是一个数组，你首先需要将数组转化为链表，然后删除链表中的重复元素，再遍历链表元素，以一个数组的形式返回。请设计一个算法解决上述任务，分析算法设计思路，计算时间复杂度, 并基于python编程实现。

e.g.  输入：head=[1, 1, 2, 3, 3]   输出：[1, 2, 3]

![image-20240502110020439](./fig/hw2q1.png)



idea：链表是有序的，其重复元素是相邻的。从头遍历链表，若当前节点值与下一个节点值相同，则跳过重复节点，否则继续向右走。

链表的编写流程参照CSDN上相关文章，在其基础上进行了学习与修改。

````python
# add your idea here
# 链表节点类定义
class NodeList:
    def __init__(self, val=None, right=None):
        self.val = val
        self.right = right

# 把数组转换成链表
def arr_to_list(arr):
    if not arr:
        return None
    result = NodeList(arr[0])
    tempt = result
    for val in arr[1:]:
        tempt.right = NodeList(val)
        tempt = tempt.right
    return result

# 删除重复元素
def Delete(head):
    tempt = head
    #只要不为None就继续
    while tempt and tempt.right:
        if tempt.val == tempt.right.val:
            # 如果重复就跳过
            tempt.right = tempt.right.right
        else:
            tempt = tempt.right
    return head

# 把链表转换回数组
def list_to_arr(head):
    result = []
    #一直循环到head为None
    while head:
        result.append(head.val)
        head = head.right
    return result

head=[1, 1, 2, 3, 3]
head=arr_to_list(head)
head=Delete(head)
head=list_to_arr(head)
print(head) 
# your algorithm time complexity is:
'T(n)=O(n)'
````

## 问题 2  

> 下面是一个经典的算法问题：
>
> - 给定包含n个整数的一个整数数组 `nums` 和一个整数目标值 `target`，请你在该数组中找出 **和为目标值** *`target`* 的那 **两个** 整数，并返回它们的**数组下标**。假设每种输入只会对应一个答案。但是，数组中同一个元素在答案里不能重复出现。你可以按任意顺序返回答案。
>
> 由于要多次查找数组中元素的位置，为了提高查询效率可以使用哈希表来存储数组中的数据，在哈希表中查询一个元素的复杂度为O(1)。 已知python中的字典是使用哈希表实现的，即使用`dict[key]`查询对应的value时间复杂度为O(1), python提供了查询字典是否包含某个key的功能：`key in dict`，时间复杂度也是O(1)

请根据上面信息，设计一个时间复杂度为O(n) 的算法，解决上述算法问题

e.g.   

输入：nums=[2,7,11,15], target=9， 输出：[0，1]

输入：nums=[3,2,4], target=6, 输出：[1,2]

输入：nums=[3,3], target=6,  输出：[0,1]


idea：利用字典记录已访问元素及其下标，边遍历边看目标差值是否已经存在。
遍历n次，每次进行一回查找与存入，题目中说每回操作是O(1)，遍历n次，就是O(n)

````python
# add your idea here
def Add(nums, target):
    # 字典
    dict = {}
    for i, num in enumerate(nums):
        # 计算差值
        index = target - num
        # 如果所需的数已经在字典里，说明找到了两个数
        if index in dict:
            # 返回两个数的下表
            return [dict[index], i]
        # 否则，把当前数字和下标加入字典中
        dict[num] = i
nums1=[2,7,11,15]
nums2=[3,2,4]
nums3=[3,3]
print("输入：nums=",nums1, "target=9，输出：",Add(nums1,9))
print("输入：nums=",nums2, "target=6，输出：",Add(nums2,6))
print("输入：nums=",nums3, "target=6，输出：",Add(nums3,6))
# your algorithm time complexity is: 
# O(n)
````

> 栈是一种常用的数据结构，编译器中通常使用栈来实现表达式求值。
>
> 以表达式 $3+5 \times 8-6$​ 为例。编译器使用两个栈来完成运算，即一个栈保持操作数，另一个栈保存运算符。
>
> 1. 从左向右遍历表达式，遇到数字就压入操作数栈；
>
> 2. 遇到运算符，就与运算符栈的栈顶元素进行比较。如果比运算符栈顶元素的优先级高，就将当前运算符压入栈；如果比运算符栈顶元素的优先级低或者相同，从运算符栈中取栈顶运算符，从操作数栈的栈顶取 2 个操作数，然后进行计算，再把计算完的结果压入操作数栈，继续比较。
>
> 下图是 $3+5 \times 8-6$  这个表达式的计算过程：

![figure](./fig/hw2q3.png)

根据上述原理，请设计一个算法完成表达式的运算，当输入为表达式字符串，返回对应的计算结果。分析算法设计思路，计算时间复杂度，并基于python编程实现

**note：**

1. 假设输入的表达式只会出现加（“+”），减 “-”， 乘“*”，除 “/” 四个运算符, 表达式中只会出现正整数
2. python中` str.isdigit()`函数可以判断字符串str是否为数字，



e.g. :
---

1. 输入：“$3+5 * 8 -6$”   输出：37

2. 输入：“$34+13*9 + 44-12/3$”  输出：191

idea：设计一个函数，将传入的字符串表达式输出整数结果。用isdigit判断是数字，压入对应的栈；若是符号，判断优先级数字是否更大，更大则压入符号的栈，否则就弹出栈顶符号与两个数字进行计算，将结果压回数字栈。比如第一个例子，是遇到-号才运算，栈顶元素是*，所以先*后+再-。
这是基于对输入字符串的从左到右的遍历，所以其时间复杂度是O(n)。


````python
# add your idea here
#由字符串返回整数
def signs(op: str) -> int:
    # 返回数字表示符号优先级，越大优先级越高
    if op in ('+', '-'):
        return 2
    elif op in ('*', '/'):
        return 4
    return 0

def operation(a: int, b: int, op: str) -> int:
    # 返回运算结果
    if op == '+':
        return a + b
    if op == '-':
        return a - b
    if op == '*':
        return a * b
    if op == '/':
        return a // b  
    return 0

def calculate_function(function: str) -> int:
    i = 0
    number = []  
    sign = []   

    while i < len(function):
        # 判断字符是否是数字
        if function[i].isdigit():
            num = 0
            # 处理多位数
            while i < len(function) and function[i].isdigit():
                num = int(function[i]) + num * 10
                i += 1
            number.append(num)
        else:
            # 确保符号栈内的操作符优先级高于当前操作符
            while sign and signs(sign[-1]) >= signs(function[i]):
                num2 = number.pop()
                num1 = number.pop()
                op = sign.pop()
                number.append(operation(num1, num2, op))
            sign.append(function[i])
            i += 1

    # 完成所有运算
    while sign:
        num2 = number.pop()
        num1 = number.pop()
        op = sign.pop()
        number.append(operation(num1, num2, op))

    return number[0]

expression1 = "3+5*8-6"
expression2 = "34+13*9+44-12/3"
print("输入：", expression1, "输出:", calculate_function(expression1))
print("输入：", expression2, "输出:", calculate_function(expression2))
# your algorithm time complexity is: 
'T(n)=O(n)'
````

## 问题 4:  

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]



idea:依次遍历数组，若当前星球向右（正数），直接加入结果数组；若向左（负数），则检查当前结果数组末尾的元素是否存在向右运动的星球，若存在则触发碰撞。碰撞后，没撞碎的星球再放入数组。
因为是依次遍历，所以是O(n)。


````python
# add your idea here
def collide(A):
    result = []
    for i in A:
        existence = True
        while result and i < 0 and result[-1] > 0:
            if abs(i) > result[-1]:
                result.pop()
                continue
            if abs(i) == result[-1]:
                existence = False
                break
            else:
                existence = False
                break
        if existence:
            result.append(i)
    return result
A1=[-3,-6,2,8, 5,-8,9,-2,1]
A2=[23,-8,9,-3,-7,9,-23,22]
print("输入：",A1,"输出:",collide(A1)) 
print("输入：",A2,"输出:",collide(A2)) 
# your algorithm time complexity is: 
'T(n)=O(n)'
````

## 问题 5 

> 给定一个无序数组nums=[9,-3,-10,0,9,7,33]，请建立一个二叉搜索树存储数组中的所有元素，之后删除二叉树中的元素“0”，再使用中序遍历输出二叉搜索树中的所有元素。

使用python编程完成上述任务，并计算时间复杂度


idea：先构建二叉搜索树，然后写插入与删除还有中序遍历函数。删除时，三个判断条件时互斥的，所以要用elif，否则会使结输出结果本来应该显示两个9，却只显示一个9.

二叉树相关代码的编写参照CSDN上相关文章，在其基础上进行了学习与修改。二叉树的代码都很像。

````python
# add your idea here
# 定义二叉搜索树
class TreeNode:
    def __init__(self, val):
        self.val = val        
        self.left = None      
        self.right = None    

# 插入新节点的函数
def insert(root, val):
    if root is None:
        return TreeNode(val)
    if val < root.val:
        root.left = insert(root.left, val)
    else:
        root.right = insert(root.right, val)
    return root

# 删除某个值的节点
def delete(root, val):
    if root is None:
        return None
    if val < root.val:
        root.left = delete(root.left, val)
    elif val > root.val:
        root.right = delete(root.right, val)
    else:
        # 找到了要删除的节点
        if root.left is None:
            return root.right
        if root.right is None:
            return root.left
        # 左右子树都不为空，找右子树中最小的节点替换
        tempt = root.right
        while tempt.left:
            tempt = tempt.left
        root.val = tempt.val
        root.right = delete(root.right, tempt.val)
    return root

# 中序遍历：左 → 根 → 右
def mid_order(root):
    if root is None:
        return []
    return mid_order(root.left) + [root.val] + mid_order(root.right)

nums = [9, -3, -10, 0, 9, 7, 33]
root = None

# 插入所有元素到二叉搜索树中
for i in nums:
    root = insert(root, i)
# 删除值为0的节点
root = delete(root, 0)

result = mid_order(root)
print("中序遍历结果：", result)
# your algorithm time complexity is:
"平均 O(log n)，最坏 O(n²)"
````

> 给定一个包含大写字母和小写字母的字符串 s ，返回 字符串包含的 **最长的回文子串的长度** 。请注意 区分大小写 。比如 "Aa" 不能当做一个回文字符串。
>

请设计一个算法解决上述问题，只需要输出最长回文子串的长度，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g. 输入： s="adccaccd"，  输出：7。 最长回文子串为："dccaccd", 长度为7


idea：使用中心扩展法，对每一个字符，分别将其作为奇数长度回文的中心和偶数长度回文的中心，通过比较回文子串长度，更新当前最长的回文子串。
对每个字符进行O(n)次扩展，所以整体时间复杂度为O(n²)。

````python
# add your idea here
def expand_around_center(s: str, left: int, right: int) -> str:
    while left >= 0 and right < len(s) and s[left] == s[right]:
        left -= 1
        right += 1
    return s[left + 1:right]

def longest(s: str) -> str:
    longest_str = ""
    for i in range(len(s)):
        # 区分奇数和偶数长度的回文串
        str1 = expand_around_center(s, i, i)  # 奇数长度
        str2 = expand_around_center(s, i, i + 1)  # 偶数长度
        
        if len(str1) > len(longest_str):
            longest_str = str1
        if len(str2) > len(longest_str):
            longest_str = str2

    return longest_str

s = "adccaccd"
result_str = longest(s)
length = len(result_str)
print("输入： s=",s, "输出：",length,"最长回文子串为：",result_str,"长度为",length)
# your algorithm time complexity is:
'T(n)=O(n²)'
````

> 沿一条长河流分散着n座房子。你可以把这条河想象成一条轴，房子是由它们在这条轴上的坐标按顺序排列的。你的公司想在河边的特定地点设置手机基站，这样每户人家都在距离基站4公里的范围内。输入可以看作为一个升序数组，数组元素的取值为大于等于0的正整数，你需要输出最小基站的数目，基站的位置。

1. 给出一个时间复杂度为$O(n$) 的算法，使所使用的基站数量最小化，分析算法设计思路，使用python编程实现
2. 证明1.中算法产生了最优解决方案。

e.g. 

输入： [1, 5, 12, 33, 34,35]  输出：基站数目为3， 基站位置为[1，12，33]



idea：采用贪婪策略，从左到右遍历数组，每次将基站设置在当前未被覆盖房子的位置加4公里处，并跳过所有能被该基站的位置+4公里能够覆盖的房子。
i只向前，不回头，每个房子最多访问一次，所以是O(n)。

````python
# add your idea here
def Place(houses):
    n = len(houses)
    i = 0
    stations = []
    while i < n:
        start = houses[i]
        stations.append(start)  
        pos=start+4
        while i < n and houses[i] <= pos + 4:
            i += 1
    return stations
houses = [1, 5, 12, 33, 34, 35]
stations = Place(houses)
print("输入：", houses, "输出：基站数目为", len(stations), "，基站位置为", stations)
# your algorithm time complexity is:
'T(n)=O(n)'
````

第2问：每一步都选择将基站放在当前未被覆盖房子坐标加4公里处，使其能够覆盖当前房子以及尽可能多的右侧房子，这是使用贪婪算法，是局部最优的。房子数组有序，基站的放置不会影响到未被覆盖房子的相对分布，所以可以看成一个个相同的小问题，每一步都是最优，那自然整体是最优。

## 问题 8  

> 给定由n个正整数组成的一个集合$S = \{a_1, a_2，···，a_n\}$和一个正整数W，设计一个算法确定是否存在S的一个子集 $K \subseteq S$, 使K中所有数之和为 $W$, 如果存在返回“True”，否则返回“False”

请设计一个时间复杂度为$O(nW)$动态规划算法，解决上述问题，分析算法的设计思路，并且基于python编程实现（不需要输出子集）。

e.g. 

输入：S = {1,4,7,3,5}， W = 11，输出：True。   因为K可以是{4,7}。



idea：使用动态规划，划分为用前i个数，能组成哪些数字和的子问题，每一行表示第i个数加进来后能表示出哪些和。dp[i][w]就是表示是否可以用前i+1个数组成目标和w。最后在这里面找，W是否在里面，在里面就输出True。
两个循环，分别是n和W+1次，所以最后为O(n)

动态规划的相关代码学习自CSDN。在其基础上学习与编写。

````python
# add your idea here
def look_for_sum(S, W):
    n = len(S)
    #初始化为全为false，行的长度为W+1，有n行
    dp = [[False] * (W + 1) for _ in range(n)]
    #什么都不选，和为0
    dp[0][0] = True 
    #S中第一个元素小于W，则将其放入dp中，其对应的本身的和可得，所以为True
    if S[0] <= W:
        dp[0][S[0]] = True 
        
    for i in range(1, n):
        current_value = S[i]  
        for w in range(W + 1):
            current_sum = w  # 当前目标和
            # 不选当前数，直接继承上一行结果
            without_current = dp[i - 1][current_sum]
            dp[i][current_sum] = without_current
            # 可以考虑选当前数
            if current_sum >= current_value:
                # 看前一行是否能组成 current_sum - current_value
                with_current = dp[i - 1][current_sum - current_value]
                # 如果选当前数也可以，则置为 True
                dp[i][current_sum] = dp[i][current_sum] or with_current
    
    return dp[n - 1][W]

S = [1, 4, 7, 3, 5]
W = 11
print("输入：S =", S, "W =", W, "输出：", look_for_sum(S, W)) 
# your algorithm time complexity is: O(nW)
````

> 给定一个n个物品的集合。物体的重量为$w_1, w_2，…、w_n$，物品的价值分别是$v_1、v_2、…v_n$。给你**两个**重量为 $c$ 的背包。如果你带了一个东西，它可以放在一个背包里，也可以放在另一个背包里，但不能同时放在两个背包里。所有权重和价值都是正整数。

1. 设计一个时间复杂度为 $O(nc^2)$ 的动态规划算法，确定可以放入两个背包的物体的最大价值。分析算法设计思路，并基于python编程实现
2. \* 修改1中的算法，输出每个背包的内容（物品对应下标）。

e.g.: 

输入 V=[1,3,2,5,8,7], W=[1,3,2,5,8,7], c=7, 输出：最大价值=14，背包装的物品为：[6] [4，3] （同一个背包中物品装入顺序对结果无影响）  


idea：使用动态规划，划分为用前i个物品，在两个背包分别使用j和k单位容量的情况下，能获得的最大总价值的子问题。dp[i][j][k]用来表示使用前i+1个物品，在背包1使用容量j、背包2使用容量k时的最大总价值。通过对比不放，放包1，放包2，将此时的最大值给到dp[i][j][k]。最终的 dp[n-1][c][c] 就是能够得到的最大总价值。
三个循环，分别是n，c+1，c+1，故总共是$O(nc^2)$ 。

背包问题是在个人对题目8的编写基础上，参照CSDN上的类似题目，进行学习与编写。

````python
# add your idea here
def two_put(W, V, c):
    n = len(W)
    # dp[i][j][k] 表示前i个物品，在两个背包容量分别为j和k时的最大价值
    dp = [[[0 for _ in range(c + 1)] for _ in range(c + 1)] for _ in range(n)]

    # 初始化第0个物品
    if W[0] <= c:
        #放包1
        dp[0][W[0]][0] = V[0]   
        #放包2
        dp[0][0][W[0]] = V[0]  
    #不放就是0    
    dp[0][0][0] = 0            

    for i in range(1, n):
        w = W[i]
        v = V[i]
        for j in range(c + 1):     
            for k in range(c + 1): 
                # 不放当前物品
                without_current = dp[i - 1][j][k]
                max_value = without_current
                # 尝试放入包1
                if j >= w:
                    with_bag1 = dp[i - 1][j - w][k] + v
                    max_value = max(max_value, with_bag1)
                # 尝试放入包2
                if k >= w:
                    with_bag2 = dp[i - 1][j][k - w] + v
                    max_value = max(max_value, with_bag2)

                dp[i][j][k] = max_value

    return dp[n - 1][c][c]

V = [1, 3, 2, 5, 8, 7]
W = [1, 3, 2, 5, 8, 7]
c = 7
result = two_put(W, V, c)
print("输入：V =", V,"W=",W,"c=",c, "输出：最大价值 =",result) 
# your algorithm time complexity is: O(nc^2)
````

## 问题 10 

> 给定两个字符串 $x[1..n]$ 和 $y[1..m]$，我们想通过以下操作将 $x$ 变换为 $y$ :
>
> **插入**：在 $x$ 中插入一个字符(在任何位置)；**删除**：从 $x$ 中删除一个字符(在任何位置)； **替换**：用另一个字符替换 $x$ 中的一个字符。
>
> 例如: $x = abcd$, $y = bcfe$，
>
> - 将 $x$ 转换为 $y$ 的一种可能方法是：1. 删除 $x$ 开头的 $a$, $x$变成 $bcd$； 2. 将 $x$ 中的字符 $d$ 替换为字符 $f$。$x$ 变成 $bcf$； 3. 在 $x$ 的末尾插入字符 $e$。$x$ 变成 $bcfe$。
>
> - 另一种可能的方法：1. 删除 $x$ 开头的 $a$,  $x$ 变成 $bcd$； 2. 在 $x$ 中字符 $d$ 之前插入字符 $f$。$x$ 变成 $bcfd$。3. 将 $x$ 中的字符 $d$ 替换为字符 $e$。$x$ 变成 $bcfe$。

设计一个时间复杂度为 $O(mn)$ 的算法，返回将 $x$ 转换为 $y$ 所需的最少操作次数。分析算法设计思路，并基于python编程实现。


idea：使用动态规划，构造二维数组dp[i][j]，表示将字符串x[1..n]转换为字符串y[1..m]所需的最小操作次数。状态转移考虑三种基本操作：删除x[i-1]的字符、插入y[j-1]的字符，以及将x[i-1]替换为y[j-1](若两字符相同则无需替换）。每步中使用delete、insert和replace记录三种操作代价，并从中选择最小值更新dp[i][j]。最后得到的dp[n][m]就是最后的结果。两个循环，分别是n+1和m+1，总共为O(mn)

部分参考CSDN,代码是在前面8，9基础上进行类比，然后依照题目写的。都是动态规划问题。

````python
# add your idea here
def transformation(x, y):
    n, m = len(x), len(y)
    # dp[i][j]表示所需的最小操作次数，n+1行，m+1列
    dp = [[0 for _ in range(m + 1)] for _ in range(n + 1)]

    # 初始化第一列：全部删除x中元素
    for i in range(n + 1):
        dp[i][0] = i
    # 初始化第一行：全部插入y中元素
    for j in range(m + 1):
        dp[0][j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            # 当前字符
            xi = x[i - 1]
            yj = y[j - 1]
            # 替换代价：相等为0，不相等为1
            if xi == yj:
                replace_cost = 0
            else:
                replace_cost = 1
            # 三种操作
            delete = dp[i - 1][j] + 1     
            insert = dp[i][j - 1] + 1   
            replace = dp[i - 1][j - 1] + replace_cost  # 替换或匹配

            # 最小操作次数
            dp[i][j] = min(delete, insert, replace)

    return dp[n][m]

x = "abcd"
y = "bcfe"
print("输入：x =", x, ", y =", y)
print("输出：最少操作次数 =", transformation(x, y))
# your algorithm time complexity is: O(mn)
````
